```python
import json

with open("data.json", "r") as f:
    data = json.load(f)

print(data)
```

PRODUCTO CON MÁS INGRESOS

```python
ingresos_prod = {}

for ventas, productos in data.items():
    for producto in productos:

        nombre = producto["producto"]
        ingresos = producto["cantidad"] * producto["precio_unitario"]

        if nombre in ingresos_prod:
            ingresos_prod[nombre] += ingresos

        else:
            ingresos_prod[nombre] = ingresos


ingreso_max = 0

for producto, ingresos in ingresos_prod.items():

    if ingresos > ingreso_max :
        ingreso_max = ingresos
        producto_mas_ingresos = producto

print(f"Producto Con Más Ingresos: {producto_mas_ingresos} - {ingreso_max:.2f} €")
```

VENTAS DE VENDEDOR CONCRETO

```python
vendedor_concreto = "Marta"

ventas_vendedor = []

for ventas, productos in data.items():
    for producto in productos:

        if producto["vendedor"] == vendedor_concreto:
            ventas_vendedor.append(producto)

print(ventas_vendedor)
```

TICKET MEDIO

```python
ingresos = 0
num_tickets = 0

for ventas, productos in data.items():
    for producto in productos:

        ingresos += producto["precio_unitario"] * producto["cantidad"]
        num_tickets += 1

ticket_medio = ingresos / num_tickets

print(f"Ticket Medio: {ticket_medio:.2f} €")
```
